```python
import pandas as pd

df = pd.read_csv("spotify.csv")
df.describe()
df.info()
df.corr()
df['dance'].corr(df['energy'])
df.head(5)
df.tail(5)
df.iloc[15:40]
df.shape
df.isnull().sum()
df.drop(columns=['a','b','c'])
df.drop(df.iloc[:10])

def outliers(df,c):
    Q1 = df[c].quantile(0.25)
    Q3 = df.quantile(0.75)
    IQR = Q3-Q1
    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR
    return df[df[c]>=lower & df[c]<=upper]

df.oultiers(df,"popu")
df.crosstab(df['time'],df["genre"])
```
